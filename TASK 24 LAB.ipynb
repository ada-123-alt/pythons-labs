```python
#our previous output display other locations which are not turly a location

df.query('Location!="Other"')\
.groupby('Location')\
['speed_mph']\
.agg(['mean','count'])\
.query('count>=10')

#.query('location='other'') this command remove other locations
#.groupby('location') this command group the data by location
#.agg(['mean','count']) this command calculate the mean and count
#.query('count>=10) this command keep only locations with at least 10 records

```

```python

ax=(
    df.query('Location!="Other"')
      .groupby('Location')
['speed_mph']\
      .agg(['mean','count'])
      .query('count>=10')
      .sort_values('mean')['mean']
      .plot(
          kind='barh',
          figsize=(12,5),
          title= 'Average coaster\
speed by location'
      )
)      
ax.set_xlabel('Average coaster speed')
plt.show()



#df.query('location!="other"') this command remove other
#group by location this command group by location
#['speed_mph'}\] this command select coaster speed
#.agg(['mean', 'count']) this command calculate average[mean average speed] and number of records[number of coasters/records.
#.query('count>=10') this command keep location with at least 10 records
#.sort_values('mean') this command select the avearsge speed column
#['mean'] this command select the average speed column
#.plot(kind='barh') this command create a horizontal bar chart
#figsize=(12,5), title='Average coaster speed by location' these commads give the chart a size and titles
#ax.set_label('Average coaster speed') this command label the xaxis
```
